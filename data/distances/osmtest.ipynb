```python
pumas = pd.read_csv("pumas.csv").set_index("GEOID")
```

```python
pumas.head()
pumas["center"] = pumas["center"].str[6:]
```

```python
pumas["center_y"] = pumas["center"].apply(lambda x: x[1:x.find(" ")])
pumas["center_x"] = pumas["center"].apply(lambda x: x[x.find(" "):])
```
