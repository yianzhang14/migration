# convert_tract_wacs_to_puma.ipynb

Similar to the tract to puma, except this does puma to migpuma aggregation.

```python
import os
import sys

import pandas as pd

sys.path.insert(0, os.path.abspath("../../.."))
import lib.io as lio
```

```python
year = 2018
```

```python
df = pd.read_csv(f"wac_puma_{year}.csv").set_index("PUMA")
df.index = df.index.astype("str").str.zfill(7)
df.head()
```

```python
puma_migpuma = lio.load_puma_migpuma("../../geometry/equivalencies/puma_migpuma.csv")
puma_migpuma.head()
```

```python
df["MIGPUMA"] = puma_migpuma.loc[df.index, "MIGPUMA"]
df.head()
```

```python
df_MIGPUMA = df.groupby("MIGPUMA").sum()
df_MIGPUMA.head()
```

```python
df_MIGPUMA.to_csv(f"wac_migpuma_{year}.csv")
```
