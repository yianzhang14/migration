# convert_tract_wacs_to_puma.ipynb

Similar to the tract to puma, except this does puma to migpuma aggregation.

```python
import pandas as pd
```

```python
YEAR = 2017
```

```python
df = pd.read_csv(f"wac_puma_{YEAR}.csv").set_index("puma")
df.index = df.index.astype("str").str.zfill(7)
df.head()
```

```python
puma_migpuma = pd.read_csv("../geometry/equivalencies/puma_migpuma.csv", dtype="str")
puma_migpuma["PUMA"] = puma_migpuma["State"] + puma_migpuma["PUMA"]
puma_migpuma["MIGPUMA"] = puma_migpuma["State"] + puma_migpuma["MIGPUMA"]
puma_migpuma = puma_migpuma.set_index("PUMA")
puma_migpuma.head()
```

```python
df.index
```

```python
df["MIGPUMA"] = puma_migpuma.loc[df.index, "MIGPUMA"]
df.head()
```

```python
df_MIGPUMA = df.groupby("MIGPUMA").sum()
df_MIGPUMA.head()
```

```python
df_MIGPUMA.to_csv(f"wac_migpuma_{YEAR}.csv")
```
