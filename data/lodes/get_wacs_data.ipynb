# get_wacs_data.ipynb

This notebook downloads the raw wac data (workplace area characteristics) for census tracts for each state and combines them into one large csv.

```python
import gzip
from io import BytesIO

import pandas as pd
import requests
```

```python
YEAR = 2017
```

```python
df = pd.DataFrame(columns=['w_geocode','C000','CA01','CA02','CA03','CE01','CE02','CE03','CNS01','CNS02','CNS03','CNS04','CNS05','CNS06','CNS07','CNS08','CNS09','CNS10','CNS11','CNS12','CNS13','CNS14','CNS15','CNS16','CNS17','CNS18','CNS19','CNS20','CD01','CD02','CD03','CD04'])
df
```

```python
states = ["AL", "AZ", "AR", "CA", "CO", "CT", "DC", "DE", "FL", "GA", 
        "ID", "IL", "IN", "IA", "KS", "KY", "LA", "ME", "MD", 
          "MA", "MI", "MN", "MS", "MO", "MT", "NE", "NV", "NH", "NJ", 
          "NM", "NY", "NC", "ND", "OH", "OK", "OR", "PA", "RI", "SC", 
          "SD", "TN", "TX", "UT", "VT", "VA", "WA", "WV", "WI", "WY"]
```

```python
"""2016 is the greatest common year between the wacs"""
for state in states:
    print(state)
    state = state.lower()
    url = "https://lehd.ces.census.gov/data/lodes/LODES7/{}/wac/{}_wac_S000_JT00_{}.csv.gz".format(state, state, YEAR)
    r = requests.get(url, allow_redirects=True)
    temp = pd.read_csv(BytesIO(gzip.decompress(r.content)))
    df = pd.concat([df, temp], axis=0, join="inner")
    
```

```python
df = df.set_index("w_geocode") 
```

```python
df.to_csv(f"wac_tracts_{YEAR}.csv")
```
